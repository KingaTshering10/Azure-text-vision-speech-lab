## Request a short explanation
An explanation in a response does not reveal or verify a model's internal reasoning.

```python
question = "There are two chickens and one duck. How many bird legs are there? Give the answer and a short calculation."
print(chat([{"role": "user", "content": question}]))
```

## Few-shot sentiment classification
The final sentence can be interpreted differently without context. Examples guide a response format; they do not establish classification accuracy.

```python
messages = [
    {"role": "system", "content": "Classify sentiment as Positive, Negative, or Ambiguous. Return only the label."},
    {"role": "user", "content": "The road is muddy and full of dirt."},
    {"role": "assistant", "content": "Negative"},
    {"role": "user", "content": "The dress is beautiful."},
    {"role": "assistant", "content": "Positive"},
    {"role": "user", "content": "This is not worth my time."},
    {"role": "assistant", "content": "Negative"},
    {"role": "user", "content": "I cannot miss this."},
]
print(chat(messages))
```
