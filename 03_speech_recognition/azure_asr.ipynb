Use a Whisper-compatible Azure deployment for the translation endpoint. A chat or TTS deployment cannot substitute for it. Configure the root `.env` first. Running request cells sends data to Azure and may incur charges. Outputs have been cleared; no new cloud run has been recorded.

```python
from pathlib import Path
import sys

ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "ai_lab.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository folder.")
sys.path.insert(0, str(ROOT))
from ai_lab import chat, image_messages, transcribe_audio, synthesize_speech
```

## Transcription
The first file uses the original filename `rr.mp3`; its language is not asserted here. Change the path to an audio file you are authorized to process.

```python
audio_dir = ROOT / "03_speech_recognition"
transcript = transcribe_audio(audio_dir / "rr.mp3")
print(transcript)
```

## Transcribe the German sample
Transcription retains the spoken language; it does not request English translation.

```python
german_transcript = transcribe_audio(audio_dir / "german.mp3")
print(german_transcript)
```

## Translate speech into English
The audio translation endpoint returns English text. It is not arbitrary target-language translation.

```python
english_translation = transcribe_audio(audio_dir / "german.mp3", translate=True)
print(english_translation)
```
